```python
df = spark.read.csv("/home/doug/project/spark/jobs/notebooks/owid-covid-data.csv", inferSchema=True, header = True)
# df.filter("total_cases > 100000").show()
# df.printSchema()
```

```python
def soma(dataframe):
    
    df = dataframe.where(f"total_cases > 0 ").groupBy("continent").agg({f"total_cases": "sum"})
    
    return df

total = soma(df)
total.explain("extended")
```

```python
# soma_udf = udf(soma, IntegerType())

def soma_udf(dataframe):
    
    df = dataframe.filter(f"total_cases > 0 ").groupBy("continent").agg({f"total_cases": "sum"})
    
    return df
spark.udf.register("udf", soma_udf)

total_cases = soma_udf(df)
total_cases.explain("formatted")
```
