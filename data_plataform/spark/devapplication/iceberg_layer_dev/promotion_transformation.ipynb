```python
dataframe = spark.sql(""" select 
                            
                            row_number() over (order by specialofferid) as promotionkey,
                            coalesce(maxqty, 9999) as maxquantity,
                            cast(discountpct as decimal(5, 2)) as discount,
                      
                        * from icebergtb.advworks_rdb.raw_dimpromotion""")
```

```python
dataframe.select(
"promotionkey",
"specialofferid",
"description",
"discount",
"type",
"category",
"minqty",
"maxquantity",
"startdate",
"enddate",
"inserted_date",
"modified_date",
"eltversion")
```
