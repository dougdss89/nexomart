```python
                                                                                  
from pyspark.sql import SparkSession

# Criar a SparkSession com configurações para o MinIO
spark = SparkSession.builder \
    .appName("Write to MinIO") \
        .config("spark.hadoop.fs.s3a.endpoint", "http://127.0.0.2:9001") \
        .config("spark.jars.packages", "org.apache.hadoop:hadoop-aws:3.3.4,com.amazonaws:aws-java-sdk-bundle:1.12.261")\
        .config("spark.hadoop.fs.s3a.access.key", "miniouser") \
        .config("spark.hadoop.fs.s3a.secret.key", "miniouser") \
        .config("spark.hadoop.fs.s3a.path.style.access", "true") \
        .config("spark.hadoop.fs.s3a.impl", "org.apache.hadoop.fs.s3a.S3AFileSystem") \
        .config("spark.hadoop.fs.s3a.connection.ssl.enabled", "false") \
        .config("spark.hadoop.fs.s3a.path.style.access", "true") \
    .getOrCreate()

# Criar um DataFrame de exemplo
data = [("Alice", 29), ("Bob", 35), ("Catherine", 24)]
columns = ["Name", "Age"]
df = spark.createDataFrame(data, columns)

# Exibir os dados para verificar o conteúdo
df.show()

# Escrever o DataFrame no MinIO em formato Parquet
output_path = "s3a://testespark/spark/"  # Substitua pelo nome do bucket configurado no MinIO
df.write.mode("overwrite").parquet(output_path)

print(f"Data written successfully to {output_path}")

# Parar a SparkSession
spark.stop()
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("HadoopClusterAccess") \
    .config("spark.hadoop.fs.defaultFS", "hdfs://localhost:9870") \
    .getOrCreate()

# Testando leitura de arquivos no HDFS
df = spark.read.text("hdfs://localhost:9870/landing/buyer/")
df.show()
```

```python
from hdfs import InsecureClient

# Endereço do NameNode (nodemaster)
namenode_ip = "172.18.0.2"
namenode_port = 9870  # Porta para a Web UI ou 9000 para RPC

# URL do NameNode
namenode_url = f"http://{namenode_ip}:{namenode_port}"

# Configuração do cliente HDFS
client = InsecureClient(namenode_url)

# Teste a conexão
print("Arquivos no diretório raiz do HDFS:", client.list('/'))
```

```python
from pyspark.sql import SparkSession

# Endereço do NameNode e ResourceManager
namenode_hdfs = "hdfs://172.18.0.2:9001"
resourcemanager_ip = "172.19.192.4"  # ResourceManager geralmente está no nodemaster
resourcemanager_port = 8088

# Criação da SparkSession
spark = SparkSession.builder \
    .appName("AccessHadoopCluster") \
    .config("spark.hadoop.fs.defaultFS", namenode_hdfs) \
    .config("spark.hadoop.yarn.resourcemanager.address", f"{resourcemanager_ip}:{resourcemanager_port}") \
    .getOrCreate()

# Testar leitura de arquivos no HDFS
df = spark.read.text("hdfs://172.19.192.1:9004/landing/buyer/")
df.show()
```
